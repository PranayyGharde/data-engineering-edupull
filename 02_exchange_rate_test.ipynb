```python
import requests

url = "https://open.er-api.com/v6/latest/GBP"
response = requests.get(url)

print(response.status_code)
```

## Read the JSON response

```python
data = response.json()
data
```

## Extract GBP → INR

```python
gbp_to_inr = data["rates"]["INR"]
gbp_to_inr
```

## Sanity check

```python
print(f"1 GBP = {gbp_to_inr} INR")
```
